### Bivariate

```python
# How does churn rate vary by Contract type?

plt.figure(figsize=(15,5))
churn_by_contract = df.groupby('Contract')['Churn'].value_counts(normalize=True).unstack() * 100
churn_by_contract.plot(kind='bar', stacked=False)
plt.title("Churn rate by contract type")
plt.xlabel("Contract Type")
plt.ylabel("Percentage of customers")
plt.xticks(rotation=0)
plt.legend(title="Churn")
plt.show()
```

##### Observation
Month to month churns around 40%, while Two year contracts churn in about 2-3%. 

```python
# How does churn rate vary by InternetService type?

plt.figure(figsize=(8,5))
churn_by_internetservice = df.groupby("InternetService")['Churn'].value_counts(normalize=True).unstack() * 100
churn_by_internetservice.plot(kind='bar',stacked=False)
plt.title("Churn rate by internet service")
plt.xlabel("Internet Service")
plt.ylabel("Percentage of customers")
plt.xticks(rotation=0)
plt.legend(title='Churn')
plt.show()
```

##### Observation:
Fiber optic customers churn the most (often 40%), DSL churns moderately, and No internet service churns the least. That last one makes sense, if you dont have internet service, you are likely just a phone only customer, less exposed to whatevers driving dissatisfaction.

Fiber Optic despite being the premium internet option, it has highest churn. Probably because of its price or service reliability issues.

```python
# How does churn rate vary by PaymentMethod?
plt.figure(figsize=(10,7))
churn_by_paymentmethod = df.groupby("PaymentMethod")['Churn'].value_counts(normalize=True).unstack() * 100
churn_by_paymentmethod.plot(kind='bar',stacked=False)
plt.title("Churn rate by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Percentage of customers")
plt.xticks(rotation=45, ha='right')
plt.legend(title='churn')
plt.tight_layout()
plt.show()
```

##### Observation
Electronic check has higher churn rate than the other three methods. Manual payment methods customers have higher churn rates than the automatic payment method

```python
# Does having techsupport affect churn rate?
plt.figure(figsize=(10,5))
churn_by_techsupport = df.groupby('TechSupport')['Churn'].value_counts(normalize=True).unstack() * 100
churn_by_techsupport.plot(kind='bar',stacked=False)
plt.title("Churn rate by Tech Support")
plt.xlabel("Tech Support")
plt.ylabel("Percentage of customers")
plt.xticks(rotation=0)
plt.legend(title="churn")
plt.show()
```

##### Observation
With no tech support, the churn percentage is about 40%.
