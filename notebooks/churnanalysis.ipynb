### hypothesis testing

```python
contract_churn = pd.crosstab(df["Contract"], df["Churn"])
```

```python
contract_churn
```

```python
from scipy.stats import chi2_contingency
```

```python
chi2, p, dof, expected = chi2_contingency(contract_churn)
```

```python
print("Chi-square statistic:", chi2)
print("P-value:", p)
print("Degrees of freedom:", dof)
```

```python
import numpy as np
n = contract_churn.to_numpy().sum()
min_dimension = min(contract_churn.shape) - 1
cramers_v = np.sqrt(chi2/(n*min_dimension))
print(f"Cramer's V: {cramers_v:.4f}")
```

### Business Interpretation
The chi-square test indicated a statistically significant association between contract type and customer churn(p<0.05). Cramér's V of 0.4101 suggests that this association is relatively strong. Therefore, contract type is likely to be an important feature for predicting customer churn.

### Internet Service
business question: Is InternetService associated with customer churn?

Step1: Hypotheses  
H0: internetservice and customer churn are independent  
H1: InternetService and customer churn are associated

```python
internet_churn = pd.crosstab(df['InternetService'],df['Churn'])
internet_churn
```

```python
chi2,p,dof, expected = chi2_contingency(internet_churn)
print("Chi-square statistic:",chi2)
print("p-value",p)
print("Degrees of freedom", dof)
```

- since p<0.05, we reject the null hypothesis
- There is statistically significant evidence that InternetService and customer churn are associated

```python
import numpy as np
n = internet_churn.to_numpy().sum()
min_dimension = min(internet_churn.shape) - 1

cramers_v = np.sqrt(chi2 / (n * min_dimension))
print(f"Cramer's V: {cramers_v:.4f}")
```

#### Business Question:
Do churned customers have different monthly charges than non churned customers?

#### Hypothesis:
Null hypothesis: The mean monthly charges is same for churned and non churned customers.  
Alternate hypothesis:The mean monthly charges are different.

```python
from scipy.stats import ttest_ind
churn_yes = df[df["Churn"] == "Yes"]["MonthlyCharges"]
churn_no = df[df["Churn"] == "No"]["MonthlyCharges"]

t_stat, p = ttest_ind(churn_yes,churn_no)

print("t-statistic:", t_stat)
print("p-value:", p)
```

#### Business interpretation
Customers who churn tend to have higher monthly charges than the customers who do not churn.

#### Business Question
Do churned customers have different tenure than non-churned customers?

#### Hypothesis:
null hypothesis: The mean tenure is the same for churned and non churned customers
alternate hypothesis: The mean tenure is different

```python
df.groupby("Churn")["tenure"].agg(["mean","median","std"])
```

```python
from scipy.stats import ttest_ind

churn_yes = df[df["Churn"] == "Yes"]["tenure"]
churn_no = df[df["Churn"] == "No"]["tenure"]

t_stat, p = ttest_ind(churn_yes,churn_no,equal_var=False)

print("t-statistic:", t_stat)
print("p-value:", p)
```

#### Business Interpretation
There is statistically significant evidence that the average tenure differs between churned and non- churned customers(p < 0.05). Customers who churn tend to have significantly shorter tenures than customers who remain, indicating that newer customers are more likely to churn.

#### Welch's t-test

```python
from scipy.stats import ttest_ind

churn_yes = df[df["Churn"] == "Yes"]["TotalCharges"]
churn_no = df[df["Churn"] == "No"]["TotalCharges"]

t_stat,p = ttest_ind(churn_yes,churn_no, equal_var=False)

print("t-statistic:", t_stat)
print("p-value",p)
```

#### Business Interpretation
There is statistically significant evidence that the mean TotalCharges differs between churned and non-churned customers.

The negative t-statistic indicates that the average TotalCharges for churned customers is lower than for non-churned customers
