```python
# How do tenure and MonthlyCharges together relate to churn
plt.figure(figsize=(10,6))
sns.scatterplot(x='tenure', y='MonthlyCharges', hue='Churn', data=df, alpha=0.5)
plt.title("Tenure vs Monthly Charges by Churn Status")
plt.xlabel("Tenure (months)")
plt.ylabel("Monthly Charges")
plt.legend(title='Churn')
plt.show()
```

##### Observation
Churn (orange points) is heavily concentrated at low tenure values (roughly 0-20 months), across a wide range of monthly charges, including mid-to-high charges ($60-$100+). Retained customers (blue) are more spread across the full tenure range, with a visible mass at high tenure (50+ months) regardless of charge level. This suggests tenure is a stronger churn driver than monthly charges alone, but customers who are both new and paying more sit in the highest risk zone, likely because they haven't yet built loyalty or seen enough value to justify the price.

```python
# What does the correlation heatmap of numeric features look like? 
plt.figure(figsize=(8,6))
numeric_cols = df[["tenure","MonthlyCharges","TotalCharges"]]
correlation_matrix = numeric_cols.corr()
sns.heatmap(correlation_matrix, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap of numeric features")
plt.show()
```

##### Observation
tenure and TotalCharges show a strong positive correlation (~0.83), which makes sense since TotalCharges accumulates over the length of a customer's tenure. MonthlyCharges and TotalCharges are moderately correlated (~0.65), while tenure and MonthlyCharges show weak correlation (~0.25),meaning how long someone stays isn't strongly tied to how much they pay per month. The high tenure-TotalCharges correlation is worth flagging as a multicollinearity concern for linear models later (e.g., Logistic Regression)<br>
I may need to drop one of these or use regularization when modeling.

```python
def tenure_bucket(tenure):
    if tenure <= 12:
        return '0-12'
    elif tenure <= 24:
        return '13-24'
    elif tenure <= 48:
        return '25-48'
    elif tenure <= 60:
        return '49-60'
    else:
        return '61-72'

df['TenureBucket'] = df['tenure'].apply(tenure_bucket)
```
